```python
import torch
from carvekit.api.high import HiInterface

# Check doc strings for more information
interface = HiInterface(object_type="object",  # Can be "object" or "hairs-like".
                        batch_size_seg=5,
                        batch_size_matting=1,
                        device='cuda' if torch.cuda.is_available() else 'cpu',
                        seg_mask_size=640,  # Use 640 for Tracer B7 and 320 for U2Net
                        matting_mask_size=2048,
                        trimap_prob_threshold=231,
                        trimap_dilation=30,
                        trimap_erosion_iters=5,
                        fp16=False)

print("Loaded")
```

```python

image_name = "sofa"
in_name = image_name+".jpg"
out_name = image_name+".png"

images_without_background = interface(['/mnt/Personal/Projects/Autofocus/test_folder/'+in_name])

cat_wo_bg = images_without_background[0]
cat_wo_bg.save('/mnt/Personal/Projects/Autofocus/Code/Object_Extraction/carvekit/output/'+out_name)
```
